## Étape 1 — Premier chargement et découverte

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Premier chargement
df = pd.read_csv('../data/dataset_secheresse_tunisie.csv', parse_dates=['date'])
print('Shape:', df.shape)
df.head()
```

```python
# Vérification des types
print(df.dtypes)
print('\nValeurs manquantes par colonne:')
print(df.isnull().sum())
```

**Observation :** Le dataset contient 122 712 lignes et 16 colonnes.  
Les colonnes `spi` et `risque_secheresse` ont des NaN pour les 3 premiers mois de chaque gouvernorat — c'est normal car le SPI nécessite un historique minimum.

## Étape 2 — Exploration des gouvernorats

```python
# Combien de gouvernorats ?
print('Gouvernorats uniques:', df['gouvernorat'].nunique())
print(df['gouvernorat'].unique())
```

```python
# Distribution des precipitations - première tentative
df['precipitations_mm'].hist(bins=50)
plt.title('Distribution des precipitations')
plt.xlabel('mm')
plt.show()

# NOTE: La distribution est très asymétrique (skewed) - beaucoup de jours à 0mm
```

```python
# Essai de visualisation par gouvernorat
moy_gouv = df.groupby('gouvernorat')['precipitations_mm'].mean().sort_values()
moy_gouv.plot(kind='barh', figsize=(10,8))
plt.title('Précipitations moyennes par gouvernorat')
plt.show()

# OBSERVATION: Le Nord (Béja, Jendouba) reçoit beaucoup plus de pluie que le Sud (Tozeur, Kébili)
```

## Étape 3 — Exploration de la variable cible

```python
# Distribution de la variable cible
print('Répartition des classes:')
print(df['risque_secheresse'].value_counts())

# PROBLÈME IDENTIFIÉ : Déséquilibre des classes
# Classe 0 très majoritaire -> nécessitera class_weight='balanced'
```

```python
# Vérification de l'indice SPI
print('Stats SPI:')
print(df['spi'].describe())

# Un SPI centré sur 0 avec écart-type proche de 1 = calcul correct ✅
print('\nMoyenne SPI:', df['spi'].mean().round(3))
print('Ecart-type SPI:', df['spi'].std().round(3))
```

## Observations et décisions pour le notebook final

1. ✅ Dataset propre avec seulement des NaN légitimes (début de période SPI)
2. ✅ Distribution précipitations très asymétrique → normalisation obligatoire
3. ✅ Déséquilibre classes → utiliser `class_weight='balanced'`
4. ✅ Forte variabilité Nord/Sud → inclure `region` comme feature
5. ✅ SPI bien calculé (moyenne ≈ 0, std ≈ 1)
